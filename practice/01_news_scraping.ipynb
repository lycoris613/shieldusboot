## 1-1. Daum 뉴스기사 제목 스크래핑하기
질문1: 아래의 url에서 뉴스기사의 링크와 제목을 출력하세요.

```python
import requests
from bs4 import BeautifulSoup

# 다음 경제 뉴스 URL
url = 'https://news.daum.net/economy'

# 요청 헤더 설정 (user-agent 없으면 접근이 막히는 경우가 있음)
req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 '
                  '(KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

print(url)

res = requests.get(url, headers=req_header)
# response를 utf-8로 encoding
res.encoding = 'utf-8'

print(type(res))
print(res.status_code)
```

```python
if res.ok:
    html = res.text
    soup = BeautifulSoup(html, 'html.parser')

    li_tags = soup.select('ul.list_newsheadline2 li')
    print(type(li_tags), len(li_tags))

    for li_tag in li_tags:
        # <a href="링크">
        a_tag = li_tag.find('a')
        link = a_tag['href']

        # <div class="cont_thumb"><strong class="tit_txt">뉴스제목</strong></div>
        strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
        title = strong_tag.text.strip() if strong_tag else '(제목 없음)'

        print(link)
        print(title)
else:
    print(f'Error Code = {res.status_code}')
```

## 1-2. Daum 뉴스기사 제목 스크래핑하기 코드를 섹션별로 처리하는 함수로 구현하기

질문2: 여러개의 section 중 하나를 선택해서 url에서 뉴스기사의 링크와 제목을 출력하는 코드를 함수로 작성하기

`print_news()` 함수에서 section name(섹션명)을 입력 받아서 `section_dict`에서 section 영문 문자열을 추출해서 url을 생성합니다. 그 이후의 코드는 1번 문제와 동일합니다.

```python
section_dict = {
    '기후/환경': 'climate', '사회': 'society', '경제': 'economy', '정치': 'politics',
    '국제': 'world', '문화': 'culture', '생활': 'life', 'IT/과학': 'tech', '인물': 'people'
}

# 함수선언
def print_news(section_name):
    # section_dict에서 영문 section명 추출 (없으면 기본값 society)
    section = section_dict.get(section_name, 'society')
    url = f'https://news.daum.net/{section}'

    print(f'======> {url} {section_name} 뉴스 <======')

    req_header = {
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 '
                      '(KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
    }

    res = requests.get(url, headers=req_header)
    res.encoding = 'utf-8'

    if res.ok:
        soup = BeautifulSoup(res.text, 'html.parser')
        li_tags = soup.select('ul.list_newsheadline2 li')

        for li_tag in li_tags:
            a_tag = li_tag.find('a')
            link = a_tag['href']

            strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
            title = strong_tag.text.strip() if strong_tag else '(제목 없음)'

            print(link)
            print(title)
    else:
        print(f'Error Code = {res.status_code}')
```

```python
print_news('경제')
print_news('사회')
```

## 2-1. Nate 뉴스기사 제목 스크래핑하기 (선택)

https://news.nate.com/recent?mid=n0100

최신뉴스, 정치, 경제, 사회, 세계, IT/과학 6개의 섹션의 뉴스를 출력하는 함수를 생성하여 스크래핑 하기 (Image, 기사제목, 기사링크). 뉴스기사의 Image가 없는 경우도 있으므로 존재 여부를 체크한 후 출력합니다.

> ⚠️ Nate 뉴스는 페이지 구조가 자주 바뀌는 편이라, 아래 코드의 CSS 선택자(`ARTICLE_SELECTOR` 등)는 실제 페이지를 크롬 개발자도구(F12)로 확인 후 상황에 맞게 조정해야 할 수 있습니다.

```python
from urllib.parse import urljoin
from IPython.display import Image, display

# 6개 섹션의 mid 코드
nate_section_dict = {
    '최신뉴스': 'n0100',
    '정치': 'n0200',
    '경제': 'n0300',
    '사회': 'n0400',
    '세계': 'n0500',
    'IT/과학': 'n0600',
}

req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 '
                  '(KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

def print_nate_news(section_name, max_count=5):
    mid = nate_section_dict.get(section_name, 'n0100')
    url = f'https://news.nate.com/section?mid={mid}'

    print(f'======> {url} {section_name} 뉴스 <======')

    res = requests.get(url, headers=req_header)
    res.encoding = 'utf-8'

    if not res.ok:
        print(f'Error Code = {res.status_code}')
        return

    soup = BeautifulSoup(res.text, 'html.parser')

    # 기사 목록 컨테이너 (실제 페이지 구조에 맞게 선택자를 조정하세요)
    article_tags = soup.select('div.mlt01, div.postSubject')  # 후보 선택자
    if not article_tags:
        # 대체: 기사 링크를 직접 찾는 방식
        article_tags = [a.parent for a in soup.select("a[href*='news.nate.com/view']")]

    for tag in article_tags[:max_count]:
        a_tag = tag.find('a', href=True)
        if not a_tag:
            continue
        link = urljoin(url, a_tag['href'])

        title_tag = tag.select_one('h2, .tit, strong') or a_tag
        title = title_tag.text.strip()
        if not title:
            continue

        print(link)
        print(title)

        # img 엘리먼트 존재 여부 체크 후 출력 (없는 기사도 있음)
        img_tag = tag.find('img')
        if img_tag and img_tag.get('src'):
            img_src = urljoin(url, img_tag['src'])
            display(Image(url=img_src))
        else:
            print('(이미지 없음)')
        print('-' * 40)
```

```python
for section in nate_section_dict:
    print_nate_news(section, max_count=3)
```

## 2-2. 하나의 네이버 웹툰과 1개의 회차에 대한 Image 다운로드 하기 (필수)

하나의 웹툰의 제목(title)과 회차번호(no), 회차의 URL(url)을 입력으로 받는 함수를 선언합니다.

`img\웹툰제목\회차번호` 디렉토리가 생성되며, 그 아래에 웹툰 image들이 다운로드 되도록 합니다.

```python
import os
from bs4 import BeautifulSoup

def download_one_episode(title, no, url):
    # referer 요청 헤더를 반드시 설정해야 한다 (없으면 이미지 다운로드가 막힘)
    req_header = {
        'referer': url,
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 '
                      '(KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
    }

    res = requests.get(url, headers=req_header)
    if not res.ok:
        print(f'Error Code = {res.status_code}')
        return

    soup = BeautifulSoup(res.text, 'html.parser')

    # 웹툰 본문 이미지는 파일명에 IMAG01 이 포함됨
    img_tags = soup.select("img[src*='IMAG01']")
    img_url_list = [img_tag['src'] for img_tag in img_tags]
    print(f'{title} {no}화 이미지 {len(img_url_list)}개 발견')

    # img/웹툰제목/회차번호 디렉토리 생성 (중간 디렉토리도 함께 생성)
    save_dir = os.path.join('img', title, str(no))
    os.makedirs(save_dir, exist_ok=True)

    for img_url in img_url_list:
        res = requests.get(img_url, headers=req_header)
        if res.ok:
            img_data = res.content
            file_path = os.path.join(save_dir, os.path.basename(img_url))
            with open(file_path, 'wb') as file:
                print(f'Writing to {file_path} ({len(img_data):,} bytes)')
                file.write(img_data)
        else:
            print(f'Error Code = {res.status_code} for {img_url}')
```

```python
download_one_episode(
    '일렉시드', 341,
    'https://comic.naver.com/webtoon/detail?titleId=717481&no=341&week=wed'
)
```

## 1. 자료구조 — 리스트 / 튜플 / 딕셔너리 / 셋

이 프로젝트의 웹스크래핑·데이터분석 코드를 이해하려면 **리스트/튜플/딕셔너리**를 반드시 알아야 합니다.
API 응답(`data/songs100.json` 등)이 대부분 **'딕셔너리들의 리스트'** 형태이기 때문입니다.

### 1-1. 리스트(list) — 순서가 있는 값의 모음
대괄호 `[]`로 만들며, 순서(인덱스)가 있고 값을 바꿀 수 있습니다(가변).

````python
fruits = ['사과', '바나나', '체리']
print(fruits)
print('개수:', len(fruits))   # len(): 요소 개수

# 인덱싱: 0부터 시작, 음수는 뒤에서부터
print('첫번째:', fruits[0])
print('마지막:', fruits[-1])
````

**슬라이싱**: `[시작:끝]` (끝 인덱스는 포함하지 않음)

````python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:3])   # 인덱스 1,2 -> [20, 30]
print(numbers[:2])    # 처음부터 인덱스 1까지 -> [10, 20]
print(numbers[2:])    # 인덱스 2부터 끝까지 -> [30, 40, 50]
print(numbers[::-1])  # step=-1 -> 리스트를 역순으로
````

**추가/삭제 — `append()` vs `extend()`**

- `append(x)` : 리스트 맨 뒤에 **x 자체를 하나의 요소로** 추가
- `extend(iterable)` : 리스트 맨 뒤에 **iterable의 요소들을 각각** 추가

````python
fruits.append('포도')          # 맨 뒤에 '포도' 하나 추가
print('append 후:', fruits)

fruits.remove('바나나')        # 값으로 삭제
print('remove 후:', fruits)

my_list = ['a']
my_list.append([1, 3, 5])      # 리스트 자체가 하나의 요소로 들어감
print('append([1,3,5]):', my_list)

my_list2 = ['a']
my_list2.extend([1, 3, 5])     # 요소 1, 3, 5 가 각각 추가됨
print('extend([1,3,5]):', my_list2)

# 반복문으로 순회 (스크래핑에서 결과 목록을 돌 때 사용)
for fruit in fruits:
    print('-', fruit)
````

### 1-2. 튜플(tuple) — 바꿀 수 없는 값의 모음
소괄호 `()`로 만들며, 한 번 만들면 값을 바꿀 수 없습니다(불변). 함수가 여러 값을 반환할 때 자주 쓰입니다.

````python
point = (3, 5)
print(point[0], point[1])

# 언패킹(unpacking): 튜플의 값을 여러 변수에 한 번에 나눠 담기
x, y = point
print('x =', x, ', y =', y)

# 함수가 여러 값을 반환하면 튜플로 묶여 나온다 -> 바로 언패킹해서 받는 패턴
def swap(a, b):
    # 두 값의 순서를 바꿔 튜플로 반환한다.
    return b, a

n1, n2 = swap(10, 20)
print('n1 =', n1, ', n2 =', n2)

# 리스트도 그대로 언패킹 가능
n1, n2 = [100, 200]
print(n1, n2)
````

**튜플/시퀀스 비교**: 파이썬은 튜플끼리 앞쪽 요소부터 순서대로 비교합니다. (생년월일로 나이 계산할 때 자주 쓰이는 트릭)

````python
print((12, 15) < (1, 1))    # False : 12 > 1 이라서 첫 자리에서 결정
print((12, 15) < (12, 17))  # True  : 첫 자리가 같아 둘째 자리(15<17)로 비교
````

### 1-3. 딕셔너리(dict) — 키:값 쌍의 모음 ⭐
중괄호 `{}`로 만들며 `키(key)`로 `값(value)`을 찾습니다.
**API 응답(JSON)의 한 건이 바로 딕셔너리**이므로 가장 중요합니다.

````python
book = {
    'title': '파이썬 입문',
    'author': '홍길동',
    'price': 25000,
}

print(book['title'])                       # 키로 접근
print(book.get('publisher', '출판사 미상'))  # get(): 키가 없어도 에러 대신 기본값 반환

book['discount'] = 22000                    # 값 추가/수정
print(book)
````

````python
for key in book.keys():          # 키만
    print('키:', key)

print('-' * 20)

for key, value in book.items():    # 키와 값 함께 (가장 많이 사용)
    print(f'{key} = {value}')
````

## 2. `*` (asterisk) 이해하기 — 언패킹(unpacking) / 패킹(packing)

`*`는 **어디에 쓰이느냐에 따라 정반대로** 동작합니다.

| 쓰이는 위치 | 역할 | 예시 |
|---|---|---|
| 함수 **호출** `f(*리스트)` | **풀기**(unpacking) — 요소를 개별 인자로 | `print(*[1,2,3])` -> `print(1,2,3)` |
| **대입문** `a, *b = ...` | **모으기**(packing) — 나머지를 리스트로 | `first, *rest = [10,20,30]` |
| 함수 **정의** `def f(*args)` | **모으기** — 인자들을 튜플로 | `args = (1, 2, 3)` |
| `**` (별 두 개) | 딕셔너리 ↔ 키워드 인자 | `f(**{'name':'홍길동'})` |

> 핵심: **괄호를 벗긴다**고 기억하면 쉽다. `f(*[1,2,3])`은 대괄호를 벗겨 `f(1,2,3)`이 된다.

````python
# 1) 함수 호출에서의 * : 리스트를 개별 인자로 '푼다'
nums = [1, 2, 3]

print(nums)     # 리스트 1개를 인자로 전달  -> [1, 2, 3]
print(*nums)    # print(1, 2, 3) 과 동일    -> 1 2 3

def add(a, b, c):
    # 세 수를 더한다.
    return a + b + c

print(add(*nums))   # add(1, 2, 3) 과 동일
````

````python
# 2) 대입문에서의 * : 나머지를 리스트로 '모은다'
first, *rest = [10, 20, 30, 40]
print('first =', first, '/ rest =', rest)

*head, last = [10, 20, 30, 40]
print('head  =', head, '/ last =', last)

a, *b, c = [1, 2, 3, 4, 5]
print('a =', a, '/ b =', b, '/ c =', c)
````

### `zip(*pairs)` — 묶인 것을 다시 푸는 전치(transpose)

앞에서 `zip(dt_list, dd_list)`로 **묶었다면**, `zip(*pairs)`는 그것을 다시 **푸는** 동작입니다.

```
('정부', 68)
('정책', 68)   ->   1번째 자리끼리 : ('정부', '정책', '주택')  -> words
('주택', 50)        2번째 자리끼리 : (68, 68, 50)            -> vals
```

````python
pairs = [('정부', 68), ('정책', 68), ('주택', 50)]

print('zip(pairs)  :', list(zip(pairs)))    # * 없이: 각 튜플이 그대로 감싸져 나옴
print('zip(*pairs) :', list(zip(*pairs)))   # * 있음: 같은 자리끼리 세로로 묶여 전치됨

words, vals = zip(*pairs)
print('words :', words)
print('vals  :', vals)
````

````python
# 실전: Counter.most_common() 결과를 그래프용 데이터로 분리하기
from collections import Counter

text = '정부 정책 주택 정부 정책 정부 민주당 주택 정부'
cnt = Counter(text.split())
print('most_common(3) :', cnt.most_common(3))

# (단어, 횟수) 쌍 리스트 -> 단어 튜플 / 횟수 튜플 로 한 번에 분리
words, vals = zip(*cnt.most_common(3))
print('words :', words)
print('vals  :', vals)

# zip 의 결과는 tuple 이므로, list 가 필요하면 map(list, ...) 로 변환
words, vals = map(list, zip(*cnt.most_common(3)))
print('list 변환 :', words, vals)
# 이렇게 분리하면 plt.bar(words, vals) 처럼 그래프에 바로 넣을 수 있다
````

### 함수 정의에서의 `*args` / `**kwargs`
함수를 **정의**할 때 붙이는 `*`는 반대로 **모으는(packing)** 역할입니다.

- `*args` : 위치 인자들을 **튜플**로 모은다
- `**kwargs` : 키워드 인자들을 **딕셔너리**로 모은다
- 호출할 때 `**딕셔너리`를 쓰면 딕셔너리가 키워드 인자로 **풀린다**

````python
def show(*args, **kwargs):
    # 가변 인자를 받아 출력한다.
    print('args   =', args)
    print('kwargs =', kwargs)

show(1, 2, 3, name='홍길동', age=20)

print('-' * 30)

info = {'name': '이순신', 'age': 30}
show(**info)   # show(name='이순신', age=30) 과 동일
````

## 3. 제어문과 함수

스크래핑 코드는 **'여러 페이지를 반복(for)하며, 조건(if)에 맞는 것만 골라, 함수(def)로 처리'**하는 구조입니다.

### 3-1. 조건문 if / elif / else
파이썬은 **들여쓰기(보통 공백 4칸)**로 코드 블록을 구분합니다.

````python
price = 28000

if price >= 30000:
    print('비쌈')
elif price >= 20000:
    print('보통')
else:
    print('저렴')

# 빈 리스트는 False 로 취급된다 (스크래핑 결과 유무 체크에 자주 사용)
results = []
if results:
    print('결과 있음')
else:
    print('결과 없음 (빈 리스트는 False)')
````

### 3-2. 반복문 for / while

````python
for name in ['BTS', 'IU', 'NCT']:
    print(name)

print('-' * 20)

# range(1, 4) -> 1, 2, 3 (끝 숫자는 포함 안 함). 페이지 순회 등에 사용
for page in range(1, 4):
    print(f'{page} 페이지 요청')
````

````python
# enumerate(): 인덱스와 값을 함께
for i, name in enumerate(['BTS', 'IU', 'NCT']):
    print(i, name)

print('-' * 20)

# while: 조건이 True인 동안 반복
count = 0
while count < 3:
    print('count =', count)
    count += 1
````

````python
# break: 반복 중단 / continue: 다음 반복으로 건너뛰기
for n in range(1, 10):
    if n == 5:
        break        # 5에서 멈춤
    if n % 2 == 0:
        continue     # 짝수는 건너뜀
    print(n)         # 1, 3 만 출력
````

### 3-3. 리스트 컴프리헨션 — 반복을 한 줄로
`[식 for 항목 in 반복대상 if 조건]` 형태로, 리스트를 간결하게 만듭니다.
실제 코드의 `[c for c in df.columns if c not in [...]]`가 이 문법입니다.

````python
# 일반 반복문
squares = []
for n in range(1, 6):
    squares.append(n * n)
print('반복문:', squares)

# 같은 결과를 컴프리헨션으로
squares2 = [n * n for n in range(1, 6)]
print('컴프리헨션:', squares2)

# if 조건 추가: 짝수의 제곱만
even_squares = [n * n for n in range(1, 11) if n % 2 == 0]
print('짝수 제곱:', even_squares)
````

### 3-4. 함수 정의와 반환

````python
def add(a, b):
    # 두 수를 더해 반환한다.
    return a + b

print(add(3, 5))
````

### 3-5. 기본값 인자 / 키워드 인자
매개변수에 기본값을 주면 호출 시 생략할 수 있습니다. **위치 인자로 전달한 값이 기본값보다 우선**합니다.

````python
def greet(name, greeting='안녕하세요'):
    # 인사말을 만든다. greeting 은 기본값이 있어 생략 가능.
    return f'{greeting}, {name}님!'

print(greet('철수'))                        # 기본값 사용
print(greet('영희', '반가워요'))             # 위치 인자로 전달 -> 기본값 대신 사용
print(greet('민수', greeting='환영합니다'))   # 키워드 인자로 전달 (이름 지정)
````

### 3-6. 람다(lambda) — 이름 없는 한 줄 함수
`lambda 인자: 표현식` 형태이며 **`return`을 쓰지 않습니다.** 정렬 기준이나 간단한 포맷 변환에 자주 쓰입니다.

````python
# 일반 함수
def double(x):
    return x * 2

# 같은 것을 람다로
double2 = lambda x: x * 2
print(double(5), double2(5))

# 천단위 콤마 포맷 (숫자 -> '1,234,567')
comma = lambda val: '{:,}'.format(int(val))
print(comma('1234567'))

# 정렬 기준(key)으로 람다 사용: 가격 높은 순
books = [{'title': 'A', 'price': 25000}, {'title': 'B', 'price': 30000}]
books_sorted = sorted(books, key=lambda b: b['price'], reverse=True)
print([b['title'] for b in books_sorted])
````

## 4. 예외처리와 파일 다루기

웹 요청은 **언제든 실패**할 수 있고, 수집한 데이터는 **파일로 저장**해야 합니다.

### 4-1. 예외처리 try / except / finally
오류가 날 수 있는 코드를 `try`에 넣고, 오류 발생 시 `except`로 처리합니다.

````python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:      # 0으로 나누는 특정 오류만 처리
        print('0으로 나눌 수 없습니다.')
        result = None
    return result

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

````python
# 여러 예외 처리 + finally (성공/실패와 무관하게 항상 실행)
def process(value):
    try:
        number = int(value)      # 숫자로 변환 시도
        print('변환 성공:', number)
    except ValueError as e:      # as e : 오류 객체를 e 에 담아 내용 확인
        print('변환 실패:', e)
    finally:
        print('처리 종료\n')    # 항상 실행 (Selenium 의 driver.quit() 위치)

process('123')
process('abc')
````

## 6. 웹 요청과 HTML 파싱

웹스크래핑의 기본 흐름: **① 요청(requests) → ② 응답 받기 → ③ 파싱(BeautifulSoup) → ④ 표로 정리(pandas)**

> 실행하려면: `pip install requests beautifulsoup4` (둘 다 파이썬 기본 라이브러리가 아닌 써드파티 라이브러리입니다)

### 6-1. requests 로 웹 요청 보내기

````python
import requests

res = requests.get('https://httpbin.org/get')
print('상태 코드:', res.status_code)                   # 200 이면 성공
print('응답 형식:', res.headers.get('Content-Type'))

# 쿼리 파라미터(검색어 등) 함께 보내기 -> URL 뒤에 ?key=value 로 붙는다
params = {'query': '파이썬', 'display': 10}
res2 = requests.get('https://httpbin.org/get', params=params)
print('\n실제 요청된 URL:', res2.url)
````

### 6-2. 응답 상태 확인과 예외처리
네트워크 요청은 실패할 수 있으므로 `try/except`로 감쌉니다. `raise_for_status()`는 200이 아니면 예외를 발생시켜 오류를 조기에 잡습니다.

````python
def fetch(url):
    # URL 에 GET 요청을 보내고 응답 객체를 반환한다. 실패 시 None.
    try:
        res = requests.get(url, timeout=10)   # timeout: 10초 넘으면 포기
        res.raise_for_status()                # 200 아니면 예외 발생
        return res
    except requests.exceptions.RequestException as e:
        print(f'요청 실패: {e}')
        return None

print('성공:', fetch('https://httpbin.org/status/200') is not None)
print('실패 처리:', fetch('https://httpbin.org/status/404'))
````

### 6-3. JSON API 응답 다루기
많은 API는 JSON을 돌려줍니다. `res.json()`으로 파이썬 딕셔너리로 변환합니다.

````python
res = requests.get('https://httpbin.org/get', params={'q': 'kpop', 'page': 1})
data = res.json()

print('타입:', type(data))
print('보낸 파라미터:', data['args'])
print('q 값:', data['args']['q'])
print('없는 키 안전 조회:', data.get('items', []))   # .get(키, 기본값) 으로 안전하게
````

### 6-4. BeautifulSoup 로 HTML 파싱
- `select(...)` : CSS 선택자로 **여러 개** 찾기 → 리스트
- `select_one(...)` : **하나** 찾기
- `find()` / `find_all()` : 태그명·속성으로 직접 검색
- 기본 파서(`html.parser`) 외에 `lxml` 파서를 쓰려면 **별도 설치**가 필요합니다.

````python
from bs4 import BeautifulSoup

html = '''
<html>
  <body>
    <ul class="book-list">
      <li class="book"><a href="/book/1">파이썬 입문</a><span class="price">25000</span></li>
      <li class="book"><a href="/book/2">데이터 분석</a><span class="price">30000</span></li>
      <li class="book"><a href="/book/3">웹 스크래핑</a><span class="price">28000</span></li>
    </ul>
  </body>
</html>
'''

soup = BeautifulSoup(html, 'html.parser')
items = soup.select('li.book')   # class가 book 인 li 요소들을 모두 선택 -> 리스트
print('찾은 개수:', len(items))
````

````python
for item in items:
    title = item.select_one('a').get_text(strip=True)
    link = item.select_one('a')['href']
    price = item.select_one('span.price').get_text(strip=True)
    print(f'{title} | {price}원 | {link}')
````

### 6-5. 파싱 결과 → DataFrame (전체 흐름 종합)
지금까지 배운 것을 모두 합칩니다: **반복 + 딕셔너리 + 예외 대비 + DataFrame + 파싱**

````python
rows = []
for item in items:
    rows.append({
        'title': item.select_one('a').get_text(strip=True),
        'price': int(item.select_one('span.price').get_text(strip=True)),
        'link': item.select_one('a')['href'],
    })

df_parsed = pd.DataFrame(rows)
print(df_parsed)

print('\n가장 비싼 책:', df_parsed.sort_values('price', ascending=False).iloc[0]['title'])

df_parsed.to_csv('parsed_books.csv', index=False, encoding='utf-8-sig')   # 한글 안깨지게
print('CSV 저장 완료')
````

> **참고 — User-Agent 헤더**: `requests.get(url, headers={'User-Agent': ...})`처럼 헤더를 지정하면 서버가 봇이 아닌 실제 브라우저 요청처럼 인식하게 할 수 있습니다 (차단 회피 목적).

## 8. Selenium 브라우저 자동화

`requests`는 JavaScript로 나중에 그려지는 내용(무한 스크롤, 로그인 후 화면 등)을 가져오지 못합니다.
**Selenium**은 실제 크롬 브라우저를 프로그램으로 조종해 이런 동적 페이지를 다룹니다.

> ⚠️ 이 섹션은 **실제 크롬 브라우저**가 필요해 자동 실행하지 않습니다. 크롬이 설치된 환경에서 셀을 직접 실행해 보세요.
> 설치: `pip install selenium` (Selenium 4.6+는 드라이버 자동 관리)

### 8-1. 드라이버 생성과 페이지 열기

````python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options

options = Options()
# options.add_argument('--headless=new')  # 창 없이 백그라운드 실행하려면 주석 해제
options.add_argument('--window-size=1200,800')

driver = webdriver.Chrome(options=options)   # Selenium 4.6+ 는 chromedriver 자동 관리
driver.get('https://quotes.toscrape.com/')
print('페이지 제목:', driver.title)
````

### 8-2. 요소 찾기 — find_element / find_elements
- `find_element(By.XXX, '값')` : 조건에 맞는 **첫 번째** 요소 하나
- `find_elements(...)` : 조건에 맞는 **모든** 요소 → 리스트 (없으면 빈 리스트)

> ⚠️ Selenium 4에서는 `find_element_by_xxx()` 같은 옛 메서드가 **삭제**되었습니다. 반드시 `find_element(By.XXX, ...)` 형태를 사용하세요.

| 방식 | 예시 | 설명 |
|---|---|---|
| `By.CSS_SELECTOR` | `'div.quote'` | CSS 선택자 (가장 많이 사용) |
| `By.ID` | `'login_field'` | id 속성 |
| `By.CLASS_NAME` | `'quote'` | class 속성 |
| `By.NAME` | `'commit'` | name 속성 |
| `By.TAG_NAME` | `'a'` | 태그 이름 |
| `By.XPATH` | `'//div[@class]'` | XPath 경로 |

````python
from selenium.webdriver.common.by import By

quotes = driver.find_elements(By.CSS_SELECTOR, 'div.quote')
print('찾은 명언 수:', len(quotes))

first = quotes[0]
text = first.find_element(By.CSS_SELECTOR, 'span.text').text
author = first.find_element(By.CSS_SELECTOR, 'small.author').text
print('명언:', text)
print('작가:', author)
````

### 8-3. 클릭과 텍스트 입력
- `.click()` : 버튼/링크 클릭
- `.send_keys('텍스트')` : 입력창에 타이핑

````python
driver.get('https://quotes.toscrape.com/login')

driver.find_element(By.CSS_SELECTOR, 'input#username').send_keys('test')
driver.find_element(By.CSS_SELECTOR, 'input#password').send_keys('secret')
driver.find_element(By.CSS_SELECTOR, 'input[type="submit"]').click()
print('현재 URL:', driver.current_url)
````

### 8-4. 대기(wait) — 매우 중요
페이지 요소는 **바로 나타나지 않을 수 있습니다.** 로딩 전에 찾으면 오류가 납니다.

- **암묵적 대기(implicit)**: 모든 요소 탐색에 '최대 N초 기다림'을 한 번만 설정
- **명시적 대기(explicit)**: 특정 조건이 만족될 때까지 기다림 (권장)

> 두 방식을 **섞어 쓰지 마세요.** 대기 시간이 예측 불가능해집니다.

````python
driver.implicitly_wait(5)   # 방법 A) 암묵적 대기

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

driver.get('https://quotes.toscrape.com/')
wait = WebDriverWait(driver, 10)   # 방법 B) 명시적 대기 (실무 권장)

first_quote = wait.until(
    EC.presence_of_element_located((By.CSS_SELECTOR, 'div.quote'))
)
print('명언이 로딩됨:', first_quote is not None)
````

### 8-5. 스크린샷과 종료
- `.save_screenshot(경로)` : 현재 화면을 이미지로 저장 (디버깅에 유용)
- `.quit()` : 브라우저 종료 (**반드시 호출**해서 자원 정리)

````python
import os
os.makedirs('img', exist_ok=True)

driver.save_screenshot('img/quotes.png')
print('스크린샷 저장 완료')

driver.quit()
print('브라우저 종료')
````

### 8-6. 예외처리 — 안정적인 자동화
요소를 못 찾거나 시간이 초과되면 예외가 납니다. `try/except/finally`로 감싸고, **finally에서 반드시 quit()**합니다 (예외가 나도 브라우저가 닫히도록).

````python
from selenium.common.exceptions import NoSuchElementException, TimeoutException

driver = webdriver.Chrome()
try:
    driver.get('https://quotes.toscrape.com/')

    rows = []
    for card in driver.find_elements(By.CSS_SELECTOR, 'div.quote'):
        rows.append({
            'text': card.find_element(By.CSS_SELECTOR, 'span.text').text,
            'author': card.find_element(By.CSS_SELECTOR, 'small.author').text,
        })

    df_quotes = pd.DataFrame(rows)
    print(df_quotes)

except (NoSuchElementException, TimeoutException) as e:
    print('요소를 찾지 못했습니다:', e)
finally:
    driver.quit()
    print('정리 완료')
````
